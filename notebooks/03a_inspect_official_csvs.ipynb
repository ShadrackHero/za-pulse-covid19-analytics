```python
import pandas as pd

CONFIRMED_URL = (
    "https://raw.githubusercontent.com/dsfsi/covid19za/master/data/"
    "covid19za_provincial_cumulative_timeline_confirmed.csv"
)
RECOVERIES_URL = (
    "https://raw.githubusercontent.com/dsfsi/covid19za/master/data/"
    "covid19za_provincial_cumulative_timeline_recoveries.csv"
)

print("pandas", pd.__version__)
```

```python
confirmed = pd.read_csv(CONFIRMED_URL)
recoveries = pd.read_csv(RECOVERIES_URL)

print("confirmed shape:", confirmed.shape)
print("recoveries shape:", recoveries.shape)
print("confirmed columns:", list(confirmed.columns))
print("recoveries columns:", list(recoveries.columns))
```

```python
print("Confirmed — first 3 rows")
display(confirmed.head(3))
print("Confirmed — last 2 rows")
display(confirmed.tail(2))
print("Recoveries — first 2 rows")
display(recoveries.head(2))
print("Recoveries — last 2 rows")
display(recoveries.tail(2))
```

```python
check_date = "01-07-2021"

c = confirmed.loc[confirmed["date"] == check_date, ["date", "YYYYMMDD", "GP", "total"]]
r = recoveries.loc[recoveries["date"] == check_date, ["date", "YYYYMMDD", "GP", "total"]]

print("Confirmed 1 Jul 2021 — expect GP 662300, total 1995556")
display(c)
print("Recoveries 1 Jul 2021 — expect GP 554529, total 1754793")
display(r)
```

```python
print("sample confirmed dates:", confirmed["date"].head(3).tolist())
print("sample recoveries dates:", recoveries["date"].head(3).tolist())
print("any ISO dates in confirmed?", confirmed["date"].astype(str).str.match(r"^\d{4}-\d{2}-\d{2}$").any())
print("any ISO dates in recoveries?", recoveries["date"].astype(str).str.match(r"^\d{4}-\d{2}-\d{2}$").any())
```

```python
print("date dtype:", confirmed["date"].dtype)
print("YYYYMMDD dtype:", confirmed["YYYYMMDD"].dtype)
print("first dates:", confirmed["date"].head(3).tolist())

mask_c = confirmed["YYYYMMDD"].astype(str).str.replace(".0", "", regex=False) == "20210701"
mask_r = recoveries["YYYYMMDD"].astype(str).str.replace(".0", "", regex=False) == "20210701"

print("\nConfirmed 1 Jul 2021")
print(confirmed.loc[mask_c, ["date", "YYYYMMDD", "GP", "total"]].to_string(index=False))

print("\nRecoveries 1 Jul 2021")
print(recoveries.loc[mask_r, ["date", "YYYYMMDD", "GP", "total"]].to_string(index=False))

print("\nISO-looking dates in confirmed?", confirmed["date"].astype(str).str.match(r"^\d{4}-\d{2}-\d{2}").any())
print("ISO-looking dates in recoveries?", recoveries["date"].astype(str).str.match(r"^\d{4}-\d{2}-\d{2}").any())
```

```python
from pathlib import Path
import pandas as pd

landing = Path.home() / "Desktop" / "za-pulse-covid19-analytics" / "data" / "landing"
landing.mkdir(parents=True, exist_ok=True)

province_cols = ["EC", "FS", "GP", "KZN", "LP", "MP", "NC", "NW", "WC", "UNKNOWN", "total"]

def to_landing(df, filename):
    out = df.rename(columns={
        "date": "DateText",
        "YYYYMMDD": "YearMonthDay",
        "UNKNOWN": "UnknownCount",
        "total": "Total",
        "source": "SourceText",
    }).copy()

    for col in ["EC", "FS", "GP", "KZN", "LP", "MP", "NC", "NW", "WC", "UnknownCount", "Total"]:
        out[col] = pd.to_numeric(out[col], errors="coerce").astype("Int64")

    out["YearMonthDay"] = out["YearMonthDay"].astype("Int64").astype(str)
    cols = [
        "DateText", "YearMonthDay",
        "EC", "FS", "GP", "KZN", "LP", "MP", "NC", "NW", "WC",
        "UnknownCount", "Total", "SourceText",
    ]
    path = landing / filename
    out[cols].to_csv(path, index=False)
    print(path)
    print("rows written:", len(out))
    return path

p1 = to_landing(confirmed, "confirmed.csv")
p2 = to_landing(recoveries, "recoveries.csv")
```
