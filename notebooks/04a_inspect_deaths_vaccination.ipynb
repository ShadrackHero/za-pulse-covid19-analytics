```python
import pandas as pd

DEATHS_URL = (
    "https://raw.githubusercontent.com/dsfsi/covid19za/master/data/"
    "covid19za_provincial_cumulative_timeline_deaths.csv"
)
VACCINATION_URL = (
    "https://raw.githubusercontent.com/dsfsi/covid19za/master/data/"
    "covid19za_provincial_cumulative_timeline_vaccination.csv"
)

print("pandas", pd.__version__)
```

```python
deaths = pd.read_csv(DEATHS_URL)
vaccination = pd.read_csv(VACCINATION_URL)

print("deaths shape:", deaths.shape)
print("vaccination shape:", vaccination.shape)
print("deaths columns:", list(deaths.columns))
print("vaccination columns:", list(vaccination.columns))
```

```python
print("Deaths — first 3 rows")
display(deaths.head(3))
print("Deaths — last 2 rows")
display(deaths.tail(2))
print("Vaccination — first 2 rows")
display(vaccination.head(2))
print("Vaccination — last 2 rows")
display(vaccination.tail(2))
```

```python
print("sample deaths dates:", deaths["date"].head(3).tolist())
print("sample vaccination dates:", vaccination["date"].head(3).tolist())
print(
    "ISO dates in deaths?",
    deaths["date"].astype(str).str.match(r"^\d{4}-\d{2}-\d{2}$").any(),
)
print(
    "ISO dates in vaccination?",
    vaccination["date"].astype(str).str.match(r"^\d{4}-\d{2}-\d{2}$").any(),
)
print("UNKNOWN column in deaths?", "UNKNOWN" in deaths.columns)
print("UNKNOWN column in vaccination?", "UNKNOWN" in vaccination.columns)
```

```python
d = deaths.loc[
    deaths["date"] == "01-07-2021",
    ["date", "YYYYMMDD", "GP", "total"],
]
v = vaccination.loc[
    vaccination["date"] == "2021-07-01",
    ["date", "YYYYMMDD", "GP", "total"],
]

print("Deaths 1 Jul 2021 — expect GP 13094, total 61029")
display(d)
print("Vaccination 1 Jul 2021 — expect GP 789010, total 3289900")
display(v)
```
